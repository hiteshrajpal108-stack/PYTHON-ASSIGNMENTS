```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV dataset
csv_path = Path("Assignment7.csv")

# Search subfolders if the CSV isn't in the notebook's current directory
if not csv_path.exists():
    matches = list(Path.cwd().rglob("Assignment7.csv"))
    if matches:
        csv_path = matches[0]
    else:
        csv_input = input("Enter the full path to Assignment7.csv: ").strip().strip('"')
        csv_path = Path(csv_input).expanduser()

        if not csv_path.is_file():
            raise FileNotFoundError(f"CSV file not found: {csv_path}")

df = pd.read_csv(csv_path)

# Display the dataset
print("Dataset:")
print(df)

# Line Plot
plt.figure(figsize=(7, 4.5))
plt.plot(df["Student"], df["Python"], marker="o")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()

# Bar Plot
plt.figure(figsize=(7, 4.5))
plt.bar(df["Student"], df["Maths"])
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.show()

# Scatter Plot
plt.figure(figsize=(7, 4.5))
plt.scatter(df["Maths"], df["Python"], s=70)
plt.title("Maths vs Python Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()

# Correlation Matrix and Heatmap
correlation = df[["Maths", "Python", "Attendance"]].corr()
print("Correlation Matrix:")
print(correlation)

plt.figure(figsize=(6.5, 5))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```
